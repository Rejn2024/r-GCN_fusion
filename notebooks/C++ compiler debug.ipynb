```python
import sys
import platform
import shutil
import os

print("OS:", platform.platform())
print("Python:", sys.executable)
print("CXX:", os.environ.get("CXX"))
for compiler in ("cl", "g++", "clang++"):
    print(f"{compiler}:", shutil.which(compiler))
```

```python
import psutil

process = psutil.Process()

for p in [process, *process.parents()]:
    try:
        if p.name().lower() == "pycharm64.exe":
            print(f'"{p.exe()}"')
            break
    except (psutil.AccessDenied, psutil.NoSuchProcess):
        continue
else:
    print("PyCharm was not found in this notebook's parent processes.")
```

```python
import os
import subprocess
import shutil
from pathlib import Path

vcvars = Path(
    r"C:\Program Files\Microsoft Visual Studio\18\Community"
    r"\VC\Auxiliary\Build\vcvars64.bat"
)

if not vcvars.exists():
    raise FileNotFoundError(vcvars)

result = subprocess.run(
    f'call "{vcvars}" >nul && set',
    shell=True,
    executable=os.environ.get("COMSPEC", "cmd.exe"),
    capture_output=True,
    text=True,
    check=True,
)

# Import Visual Studio's environment into this notebook process.
for line in result.stdout.splitlines():
    key, separator, value = line.partition("=")
    if separator and key and not key.startswith("="):
        os.environ[key] = value

print("Compiler:", shutil.which("cl"))
```
